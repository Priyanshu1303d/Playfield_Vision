```python
#Move the best Model to the models folder
source_path = "training_results/football_v5_run/weights/best.pt"

destination_path = "models/best_yolov5.pt"

import shutil 

if os.path.exists(source_path):
    os.makedirs("models", exist_ok=True) # Create 'models' folder if missing
    shutil.copy(source_path, destination_path)
    print(f"✅ Success! Model moved to: {destination_path}")
else:
    print(f"❌ Error: Could not find file at {source_path}")

```
